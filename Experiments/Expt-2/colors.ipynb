# RGB (Red, Green, Blue)
- It's an additive color model (used in screens, projectors)
- Adding all three gives white
- Removing all gives black

# CMY (Cyan, Magenta, Yellow)
- It's a subtractive color model (used in printing, pigments) 
- Adding all gives black
- Removing all gives white

## Complementary Relation
- Red -> Cyan
- Green -> Magenta
- Blue -> Yellow

```python
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
bgr = cv.imread("rgb.png", cv.IMREAD_COLOR)
rgb = bgr[:, :, ::-1]
plt.figure(figsize=[20, 20])
plt.subplot(121); plt.imshow(bgr); plt.title("BGR")
plt.subplot(122); plt.imshow(rgb); plt.title("RGB")
```

```python
ycm = cv.imread("cmy.png", cv.IMREAD_COLOR)
cmy = ycm[:, :, ::-1]
plt.figure(figsize=[20, 20])
plt.subplot(121); plt.imshow(cmy); plt.title("CMY")
plt.subplot(122); plt.imshow(ycm); plt.title("YCM")
```

```python
r, g, b = cv.split(rgb)
plt.figure(figsize=[20, 5])
plt.subplot(141); plt.imshow(r); plt.title("Red")
plt.subplot(142); plt.imshow(g); plt.title("Green")
plt.subplot(143); plt.imshow(b); plt.title("Blue")
plt.subplot(144); plt.imshow(rgb); plt.title("RGB")
```
