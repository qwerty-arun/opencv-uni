```python
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
light = cv.imread("../../assets/light.png", cv.IMREAD_COLOR);
light = light[:, :, ::-1]
plt.imshow(light)
```

```python
r, g, b = cv.split(light)
plt.figure(figsize=[20, 5])
plt.subplot(141);plt.imshow(r,cmap="gray");plt.title("Red");
plt.subplot(142);plt.imshow(g,cmap="gray");plt.title("Green");
plt.subplot(143);plt.imshow(b,cmap="gray");plt.title("Blue");
light_merge = cv.merge((r, g, b))
plt.subplot(144);plt.imshow(light_merge);plt.title("Merged")
```

## Hue, Saturation and Value
- `H` ranges from 0 to 179. Represents the full spectrum of colors, where 0 and 179 both correspond to shades of red.
- `S` ranges from 0 to 255. This represents the intensity or purity of the color. '255' being fully saturated (pure color) and '0' being completely desaturated(grayscale).
- `V` ranges from 0 to 255. Represents brightness. '255' represents fully bright and 0 being completely black.

```python
light_hsv = cv.cvtColor(light, cv.COLOR_RGB2HSV)
h, s, v = cv.split(light_hsv)
light_merge_hsv = cv.merge((h, s, v))
plt.figure(figsize=[20, 5])
plt.subplot(141);plt.imshow(h,cmap="gray");plt.title("Hue");
plt.subplot(142);plt.imshow(s,cmap="gray");plt.title("Saturation");
plt.subplot(143);plt.imshow(v,cmap="gray");plt.title("Value");
plt.subplot(144);plt.imshow(light_merge_hsv);plt.title("Merged")
```
